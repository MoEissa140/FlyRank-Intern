## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import numpy as np

# Load the starter anonymized CSV
df = pd.read_csv("/content/content_refresh_anonymized.csv")

# Metric 1: Total pages vs Declining pages proportion
total_pages = len(df)
declining_pages = (df["trend_direction"].str.lower() == "down").sum()
decline_rate = (declining_pages / total_pages) * 100

# Metric 2: High visibility pages (impressions >= 500) that are decaying
high_vis_decay = df[(df["impressions_90d"] >= 500) & (df["trend_direction"].str.lower() == "down")]

# Metric 3: Stale pages (not updated in >= 180 days) with ongoing demand (impressions >= 500)
stale_high_demand = df[(df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)]

print(f"1. Total Pages Analyzed: {total_pages:,} | Overall Decline Rate: {decline_rate:.1f}%")
print(f"2. High-Visibility Decaying Pages (Impressions >= 500 & Trend == 'down'): {len(high_vis_decay):,}")
print(f"3. Stale High-Demand Pages (Unupdated >= 180 days & Impressions >= 500): {len(stale_high_demand):,}")
```

### Key Insights from the Starter Dataset:
1. **Widespread Traffic Decay:** Out of 30,000 total pages analyzed, **54.2%** are currently experiencing a downward traffic trend (`trend_direction == 'down'`).
2. **Massive High-Value Opportunity:** There are **9,961 high-visibility pages** (generating ≥ 500 impressions in 90 days) that are actively declining. This represents a huge, high-priority pool where prioritization and refresh recommendations can prevent massive traffic loss.
3. **Immediate Low-Hanging Fruit:** **17 pages** have active demand (≥ 500 impressions) despite not being updated for over 6 months (≥ 180 days), making them clear candidates for quick-win content updates.

These numbers confirm that traffic decay is widespread across the inventory and that an ML-based ranking queue is required to prioritize which of the 9,961 declining pages the editorial team should review first.
