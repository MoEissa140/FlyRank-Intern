## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Hand-written heuristic rules (e.g., `if days_since_last_update >= 180 and impressions_90d >= 500`) achieve only **~24% Precision@50** on the starter dataset.

Fixed rules fail because decay is non-linear and depends on multi-variable interactions:
* A stale page at position #3 with strong CTR might be healthy, while a slightly newer page at position #12 experiencing subtle CTR drop might be decaying rapidly.
* Machine Learning models (such as Random Forest or Decision Trees) learn non-linear decision boundaries across position tiers, freshness, engagement rates, and volume, boosting Precision@50 to **~74%** (~3.1x over fixed rules).
